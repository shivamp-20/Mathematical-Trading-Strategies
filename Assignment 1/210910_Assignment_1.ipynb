```python
#np.std(data,ddof=1)

def std_dev(data):
    # Get number of observations
    n = len(data)
    # Calculate mean
    mean = sum(data) / n
    # Calculate deviations from the mean
    deviations = sum([(x - mean)**2 for x in data])
    # Calculate Variance & Standard Deviation
    variance = deviations / (n - 1)
    s = variance**(1/2)
    return s
```

```python
def sharpe_ratio(return_series, N, rf):
    mean = (return_series.mean() * N) - rf #this mean is excess return over the risk free rate
    sigma = return_series.std() * np.sqrt(N) #This calculates annualized volatility.
    return mean / sigma

N = 255 #255 trading days in a year
rf = 1.4/100 #1% risk free rate

returns = data.pct_change().dropna()

sharpes = returns.apply(sharpe_ratio, args=(N,rf,),axis=0)

#sharpe ratio = (Portfolio return - Risk-free return)/Standard deviation (risk) 
#higher the sharpe ratio, better the risk-adjusted return of the portfolio.
sharpes
```

```python
#data = yf.download(ticker_symbols, start='2010-01-01', end='2023-01-01')

#it tells you the worst loss an investor would have experienced by buying at a peak and holding until the lowest point before recovery.
#That means an investor who bought at the peak (120) and sold at the bottom (90) would have lost 30%.
def calculate_max_drawdown(prices):
    cumulative_returns = prices / prices.iloc[0]
    previous_peaks = cumulative_returns.cummax()
    drawdowns = (cumulative_returns - previous_peaks) / previous_peaks
    max_drawdown = drawdowns.min()
    return max_drawdown

# Calculate maximum drawdown for each equity
max_drawdowns = adj_close_prices.apply(calculate_max_drawdown, axis=0)

# Display the maximum drawdown for each equity
print(max_drawdowns)     
```

```python
# The Sortino Ratio is very similar to the Sharpe Ratio, but with one major difference:
# Sharpe Ratio penalizes all volatility (both upside and downside).
# Sortino Ratio penalizes only downside volatility (bad volatility).
# This makes the Sortino Ratio a better measure when you only care about harmful fluctuations.

def calculate_sortino_ratio(prices, risk_free_rate=0.0):
    returns = prices.pct_change().dropna()
    downside_returns = np.where(returns < risk_free_rate, returns, 0)
    downside_deviation = np.std(downside_returns)
    
    expected_return = np.mean(returns)
    sortino_ratio = (expected_return - risk_free_rate) / downside_deviation
    return sortino_ratio

# Calculate Sortino ratio for each equity
sortino_ratios = adj_close_prices.apply(calculate_sortino_ratio, risk_free_rate=0.02)

# Display the Sortino ratio for each equity
print(sortino_ratios)
```
